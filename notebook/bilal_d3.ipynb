```python
df["age_band"] = pd.cut(
df["age"],bins=[0,35,55,120]   ,labels=("youth","middle-aged","senior")
)
```

```python
df["age_band"].value_counts()
```

create a bmi category features

```python
df["bmi_category"] = pd.cut(
df["bmi"],bins=[0,18.5,25, 30,100],labels=("underweight","normal","overweight","obese"))
```

```python
df['bmi_category'].value_counts()
```

```python
df["glucose_band"] = pd.cut(
df["glucose"],bins=[0,100,126,300],labels=("normal","prediabetic","diabetic"))
```

```python
df['glucose_band'].value_counts()
```

create an insulin to glucose ratio feature

```python
df['insulin_glucose_ratio']=df['insulin']/df['glucose']
```

```python
df['insulin_glucose_ratio'].head(10)
```

```python
df['insulin_glucose_ratio'].describe()
```

check the ratio for impossible values

```python
np.isinf(df['insulin_glucose_ratio']).sum()
```

create a pregnancies-per-year-of-age-feature

```python
df['pregnancies_age_rate'] = df['pregnancies']/df['age']
```

```python
df['pregnancies_age_rate'].describe()
```

create a days-since-last-checkup feature

```python
reference_date=df['last_checkup_date'].max()
```

```python
reference_date
```

```python
df['since_checkup']=(reference_date-df['last_checkup_date']).dt.days
```

```python
df['since_checkup'].head(10)
```

create a categorical comination feature

```python
df['activity_smoker']=df['physical_activity'] + "/" +df['smoker']
```

```python
df['activity_smoker'].head()
```

```python
df['activity_smoker'].value_counts
```

#EXPLORATORY ANALYSIS

```python
numeric_cols=['age','glucose','blood_pressure','skin_thickness','insulin','bmi','diabetes_pedigree','pregnancies']
df.groupby('outcome')[numeric_cols].mean()
```

```python
df.head()
```

run the same comparison  on your engineered numeric features

```python
engineered_numeric = ['insulin_glucose_ratio','pregnancies_age_rate','since_checkup']
df.groupby('outcome')[engineered_numeric].mean()
```

rank everything together

```python
group_means=df.groupby('outcome')[numeric_cols+engineered_numeric].mean().T
group_means['gap']=(group_means[1]-group_means[0]).abs()
group_means.sort_values('gap',ascending=False)
```

seprate the featurees from the target

```python
feature_cols =['age','pregnancies','glucose','blood_pressure','skin_thickness','insulin','bmi','diabetes_pedigree','bmi_category','glucose_band']
x=df[feature_cols]
y=df['outcome']
```

encode your banded features

```python
x=pd.get_dummies(x,drop_first=True)
#drop_first is used for after new column add before column removing function 
#get dummies is  used for value to numerical value function  
x.shape
```

split into trainning and test sets

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(
    x,y, test_size=0.2, random_state=42, stratify=y)
```

BUILDING THE BASELINE

import your model and scaler

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
```

scale your features

```python
scaler = StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled= scaler.transform(x_test)
```

```python
x_test_scaled
```

fit the baseline model

```python
model = LogisticRegression(max_iter=1000)
model.fit(x_train_scaled, y_train)                           
```

predict on the test set

```python
y_pred = model.predict (x_test_scaled)
y_pred[:10]
```

eyeball the first ten prediction

```python
pd.DataFrame({"actual":y_test.values[:10],
             "predicted":y_pred[:10]})
```

compute test accuracy

```python
from sklearn.metrics import accuracy_score
test_acc=accuracy_score (y_test,y_pred)
print("test_accuracy =",test_acc)
```

```python
train_acc=accuracy_score(y_train,model.predict(x_train_scaled))
print(train_acc,test_acc)
print(train_acc-test_acc)
```
